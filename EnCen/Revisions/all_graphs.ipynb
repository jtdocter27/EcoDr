### All Difference Scores Concatenated


```python
### Loop through the folder
diff_score_directory = '/home/anna/Documents/EcoGenoRisk_Paper_Revisions/BioPesticide_Difference_Scores_per_Biome'
files_sorted = sorted(os.listdir(diff_score_directory))
iterations = 5

for i in range(0, len(files_sorted), iterations):
    batch_files = files_sorted[i:i + iterations]
    results = []
    for file in batch_files:
        file_path = os.path.join(diff_score_directory, file)
        score_df = pd.read_csv(file_path, sep='\t')
        # print(score_df.shape)

        filename = file
        biome_name = re.search(r'And(.*?)_Metagenome', filename)
        if biome_name:
            extracted = biome_name.group(1)
            # print(extracted)
            extracted = extracted.replace('_', " ")
        score_df['Biome'] = extracted
        
        score_df["JGI Bin ID"] = score_df["Organisms Compared to Synbio"].str.replace('_matches.tsv', " ", regex=False)
        score_df = score_df.drop(columns="Organisms Compared to Synbio")
        score_df = score_df.drop(score_df.columns[[0]], axis = 1)
        results.append(score_df)

    bio_name = file.split('.', 1)[0]
    bio_name = bio_name.replace('_', " ")
    print(bio_name)
    chunk_dataframe = pd.concat(results, axis = 0)
    # chunk_dataframe.to_excel('Difference Scores Concatenated' + str(counter) + '.xlsx')
    # counter += 1
    display(chunk_dataframe)
    # print(chunk_dataframe.shape)

### Read in each file 
### Take the top 10 difference scores
### Assign the biome to the difference scores
### Put into a list 
### concatane per synbio org
```
