```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm, gamma, lognorm, weibull_min, gaussian_kde, skew, genpareto
from matplotlib.ticker import FuncFormatter
import os
```

```python
as_all = pd.read_excel('Biome Synbio Top Match EC Comparison.xlsx')
as_all53 = as_all[as_all['Percentage'] > 0]
as_lost53 = pd.read_excel('/home/anna/Desktop/JD_Niche_OverLap (Git)/Niche_JD/Eco_V2/EnCen/Figures for paper/Lost_Functions.xlsx')
as_present53 = pd.read_excel('Only Present Enzyme Biome Synbio Top Match EC Comparison.xlsx')
added_EC53 = as_all53[(as_all53['Synbio Presence/Absence'] == 1) & (as_all53['Top Match Presence/Absence'] == 0)]
as_overlap53 = as_all53[(as_all53['Synbio Presence/Absence'] == 1) & (as_all53['Top Match Presence/Absence'] == 1)]
as_lost53
```

```python
bin_edges = range(0, 101, 5)
sns.histplot(x='Percentage', bins = bin_edges, data = added_EC53, color='black', label = 'Present Functions', kde = True, alpha = 0.5)
sns.histplot(x='Percentage',bins=bin_edges, data = as_overlap53, color = 'red', label='Added Functions', kde= True, alpha = 0.5)
plt.show()
```

```python
bin_edges = range(0, 101, 5)
sns.histplot(x='Percentage', bins = bin_edges, data = as_lost53, color='blue', label = 'Displaced Functions', alpha = 0.8)
sns.histplot(x='Percentage',bins=bin_edges, data = added_EC53, color = 'red', label='Added Functions', alpha = 0.5)
plt.xlabel('Bin Percentage')
plt.ylabel('EC Count')
plt.title('Bin Completeness = 53.45%')
plt.legend()
plt.savefig('/home/anna/Desktop/A&L_53%.svg', format='svg')
plt.show()
```

```python
bin_edges = range(0, 101, 5)

plt.hist([as_lost53['Percentage'], as_overlap53['Percentage'],added_EC53['Percentage']], 
         bins=bin_edges, 
         label=['Displaced Functions', 'Overlapped Functions','Added Functions'],
         color=['blue', 'purple', 'red'],
         alpha=0.5,
         stacked=True)

plt.xlabel('Bin Percentage')
plt.ylabel('EC Count')
plt.title('Bin Completeness = 53.45%')
plt.ylim(0, 80)
plt.yticks(range(0, 81, 10))
plt.legend()
plt.savefig('/home/anna/Desktop/A&L_53%.svg', format='svg')
```
