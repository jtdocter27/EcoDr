#### These are hisotrgrams of difference scores for the supplementals of EcoGenoRisk Paper

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
```

### Kleb

```python
##For Klebsiella Validation
bulk_soil = pd.read_csv('/Users/johndocter/Documents/Research/EnCen Paper/New Results/Klebsiella /Bulk Soil/Bulk_Soil.txt', sep='\t')
bulk_soil['Biome'] = 'Agricultural Bulk Soil'
# diff_score1
endo = pd.read_csv('/Users/johndocter/Documents/Research/EnCen Paper/New Results/Klebsiella /Endosphere/Absolute_Difference_Comparison_Score.txt', sep='\t')
endo['Biome'] = 'Soil Endosphere'
# diff_score2
rhizo = pd.read_csv('/Users/johndocter/Documents/Research/EnCen Paper/New Results/Klebsiella /Rhizosphere/Absolute_Difference_Comparison_Score.txt', sep='\t')
rhizo['Biome'] = 'Soil Rhizosphere'
rhizo
```

```python
frames = [bulk_soil, endo, rhizo]
    # frames
combined= pd.concat(frames)
combined_sorted = combined.sort_values(by='Biome')
combined_sorted = combined_sorted.drop('Unnamed: 0', axis=1)
combined_sorted['Metagenome Bin ID'] = combined_sorted['Metagenome Bin ID'].str.replace('_matches.tsv', '')
combined_sorted
```

```python
plt.figure(figsize=(15, 9))
sns.barplot(data = combined_sorted, x = 'Metagenome Bin ID', y = 'Difference Score', hue='Biome', palette='Blues')
```
